#### SelectKbest analysus:<br>
#### <br> "median_income" has a much higher score than all other variables. This variable playes the most important role of housing prices. Location-related variables such as ocean proximity and distance to the nearest city also appear to play a major role. These varialbes: "households","total_bedrooms","population", "ocean_proximity_ISLAND" - are less important.
#### Comparison with RFE results<br>The SelectKBest results show a clearer result of feature importance in my opinion, with very high scores for median_income, indicating a very strong relationship with the target variable. Location-related features such as ocean_proximity_INLAND and distance_to_nearest_city also received high scores, meaning that geographic factors play a significant role in determining housing prices. There are also some differences between the methods. RFE selected population and total_bedrooms, while SelectKBest did not consider them among the most important features. Instead, SelectKBest selected additional ocean proximity categories such as <1H OCEAN and NEAR BAY.<br>
