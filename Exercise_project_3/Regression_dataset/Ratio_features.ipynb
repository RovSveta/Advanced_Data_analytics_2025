```python
import pandas as pd
# load df:

df = pd.read_csv("housing_modified.csv")
```

```python
# Bedrooms compared to total rooms
df["bedrooms_to_rooms_ratio"] = df["total_bedrooms"] / df["total_rooms"]

# Households compared to population
df["households_to_population_ratio"] = df["households"] / df["population"]

# View the new features
df.head()
```

```python
# I will drop "total_bedrooms" and "households" as they are now represented in the new features:
df = df.drop(["total_bedrooms", "households"], axis=1)
```

```python
df.head()
```

```python
# now i will run again selectKBest tool to see if the new features are selected as important or not.
from sklearn.feature_selection import SelectKBest, f_regression

X = df.drop("median_house_value", axis=1)
y = df["median_house_value"]
```

```python
# Apply SelectKBest:
# same number RFE selected - 7 so i can compare results.
selector = SelectKBest(score_func=f_regression, k=7)

X_selected = selector.fit_transform(X, y)
```

```python
# Get feature scores
#the results are very similar 
scores = pd.DataFrame({
    "Feature": X.columns,
    "Score": selector.scores_
})

scores = scores.sort_values(by="Score", ascending=False)

print(scores)
```

```python
# Save the updated dataframe with ratio features
df.to_csv("housing_with_ratio_features.csv", index=False)
```
