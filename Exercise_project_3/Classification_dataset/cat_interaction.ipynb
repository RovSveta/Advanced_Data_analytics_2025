```python
import pandas as pd

df = pd.read_csv("mushroom_encoded.csv")

df.head()
```

```python
# Gill structure interaction
df["gill_spacing_attachment"] = (
    df["gill-spacing_d"] * df["gill-attachment_p"]
)

# Stem structure interaction
df["stem_structure"] = (
    df["stem-width"] * df["stem-height"]
)

# Cap appearance interaction
df["cap_surface_color"] = (
    df["cap-surface_s"] * df["cap-color_r"]
)

print(df[[
    "gill_spacing_attachment",
    "stem_structure",
    "cap_surface_color"
]].head())
```

```python

from sklearn.feature_selection import mutual_info_classif

# Define features and target
X = df.drop("class", axis=1)
y = df["class"]

# Calculate Mutual Information scores
mi_scores = mutual_info_classif(X, y)

# Convert to dataframe for easier reading
mi_importance = pd.DataFrame({
    "Feature": X.columns,
    "MI Score": mi_scores
}).sort_values(by="MI Score", ascending=False)

print("\nTop important features after interaction:")
print(mi_importance.head(10))
```
