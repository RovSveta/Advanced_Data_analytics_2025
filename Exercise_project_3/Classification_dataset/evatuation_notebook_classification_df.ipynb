### Redundancy management and dimensionality reduction. PCA – Principal Component Analysis
#### <br>The PCA results show that the first 10 principal components explain about 30% of the total variance in the dataset. Each individual component explains only a small portion of the variance (around 2–5% per component). This indicates that the information in the mushroom dataset is spread across many variables, which is common for datasets with many one-hot encoded categorical features. Because of this, PCA is not very efficient at compressing the dataset into a small number of components. (Chatgpt helped to conlude that).
