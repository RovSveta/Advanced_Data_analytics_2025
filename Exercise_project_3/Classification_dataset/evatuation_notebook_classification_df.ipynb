#### Decision Tree results.

#### SHAP results.<br>Both SHAP analysis and the Decision Tree model show similar results for the mushroom dataset. In both methods, stem-width appears as the most important feature. Several other features are also important in both methods, such as gill-spacing_d, stem-height, stem-color_w, cap-surface_s, and gill-attachment_p. There are also some small differences. For example, SHAP highlights features like gill-color_y and gill-color_w, while the Decision Tree model identifies ring-type_z and cap-color_r as important.
###  Interaction features. Categorical interactions
#### Since i use mushroom dataset that is already encoded, the best categorical interactions are between important features that already identified with SHAP and the Decision Tree. would be good to combine variables that are biologically related and important for classification. For example, gill spacing and attachment together describe gill structure, stem width and height together describe overall stem size and a cap surface and color together describe cap appearance, a key mushroom characteristic.<br>
<br>
#### The interaction feature added some additional useful information, but it did not change the overall feature importance ranking. The dataset structure remains similar, with stem characteristics and cap features being the most important factors in mushroom classification. The newly created interaction feature stem_structure appears as the second most important variable but as it contains values of first and third important variables it will just add redundancy to this df.
