```python
import pandas as pd

# boxplot version
import seaborn as sns
import matplotlib.pyplot as pl

df = pd.read_csv("housing_trimmed_variables.csv")

# in this dataset i have already used quantile filtering, so i can use it further. 

# for easier visualizations, we'll split the continuous price into categories
df['binned_price'] = pd.cut(df['median_house_value'], bins=5, labels=['Low', 'Below Average', 'Average', 'Above Average', 'High'])
```

```python
df.head()
```

```python
# checking the outliers for large buildings and price
# it seems the largest buildings are actually not the most expensive ones, but they are expensive.
sns.scatterplot(data=df, x="median_income", y="median_house_value")
```

```python
# compare binned price to grade
plt.figure(figsize=(10, 6))
sns.boxplot(x='binned_price', y='ocean_proximity_INLAND', data=df, palette='coolwarm')
plt.title('Price (binned) vs ocean_proximity_INLAND')
plt.xlabel('Price (binned)')
plt.ylabel('ocean_proximity_INLAND')
plt.show()
```
